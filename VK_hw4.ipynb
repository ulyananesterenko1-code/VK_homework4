# Задача 1. Восстановление бинарного дерева из массива

## Условие
Дан массив, представляющий бинарное дерево (например, в виде уровнями — level-order). Необходимо восстановить структуру бинарного дерева.

**Что нужно сделать:**
Реализуйте функцию, которая строит дерево по переданному массиву. Объясните, как определяется связь родитель–потомок и какая структура данных используется для хранения узлов.

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(arr, i=0):
    if i >= len(arr) or arr[i] is None:
        return None

    root = TreeNode(arr[i])
    root.left = build_tree(arr, 2 * i + 1)
    root.right = build_tree(arr, 2 * i + 2)

    return root

def print_level_order(root):
    if not root:
        return []
    queue = [root]
    result = []
    while queue:
        node = queue.pop(0)
        if node:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            result.append(None)
    while result and result[-1] is None:
        result.pop()
    return result

arr = list(map(int, input("Введите элементы массива через пробел: ").split()))

root = build_tree(arr)
print("Дерево успешно построено.")
print("Проверка (вывод по уровням):", print_level_order(root))
```

Для массива, представляющего дерево, связь родитель-потомок вычисляется по формулам:
* Левый потомок: 2 * i + 1
* Правый потомок: 2 * i + 2

Для хранения узлов используется класс TreeNode, который содержит значение val и указатели на left и right. Построение происходит рекурсивно: мы создаем узел и вызываем функцию для его левого и правого потомков.

# Задача 2. Симметричное бинарное дерево

## Условие
Дано бинарное дерево. Необходимо определить, является ли оно симметричным относительно своей середины.

**Что нужно сделать:**
Реализуйте рекурсивную или итеративную проверку зеркальности левого и правого поддеревьев. Поясните, какие пары узлов сравниваются и почему алгоритм работает корректно.

```python
def is_symmetric(root):
    if not root:
        return True

    def is_mirror(left, right):
        if not left and not right:
            return True

        if not left or not right:
            return False

        return (left.val == right.val) and \
               is_mirror(left.left, right.right) and \
               is_mirror(left.right, right.left)

    return is_mirror(root.left, root.right)

arr = list(map(int, input("Введите элементы массива через пробел (None для пустых): ").split()))
arr = [int(x) if x != 'None' else None for x in arr]

root = build_tree(arr)
result = is_symmetric(root)
print(f"Является ли дерево симметричным: {result}")
```

Мы сравниваем пары узлов: левый потомок левого поддерева с правым потомком правого поддерева, и наоборот. Если значения совпадают и их структура зеркальна, дерево симметрично. Алгоритм корректен, так как он проверяет все уровни дерева рекурсивно.

# Задача 3. Поиск минимальной глубины бинарного дерева

## Условие
Дано бинарное дерево. Необходимо найти его минимальную глубину (расстояние от корня до ближайшего листа).

**Что нужно сделать:**
Реализуйте решение с использованием BFS или DFS. Объясните, почему поиск в ширину (BFS) позволяет корректно найти минимальную глубину.

```python
def min_depth(root):
    if not root:
        return 0

    if not root.left:
        return 1 + min_depth(root.right)
    if not root.right:
        return 1 + min_depth(root.left)

    return 1 + min(min_depth(root.left), min_depth(root.right))

arr = list(map(int, input("Введите элементы массива через пробел: ").split()))
root = build_tree(arr)
print(f"Минимальная глубина дерева: {min_depth(root)}")
```

BFS обходит дерево по уровням (сверху вниз). Первый лист, который мы встретим при таком обходе, гарантированно будет находиться на минимальной глубине. Мы просто останавливаем поиск, как только дошли до него.

# Задача 4. Поиск произведения максимального и минимального элементов

## Условие
Дано бинарное дерево поиска (BST), представленное в виде массива. Необходимо найти произведение минимального и максимального значений.

**Что нужно сделать:**
Восстановите BST (если требуется) и найдите минимальный и максимальный элементы, используя свойства BST. Поясните, почему поиск крайних элементов выполняется за O(h), где h — высота дерева.

```python
def find_min_max_product(root):
    if not root:
        return 0

    current_min = root
    while current_min.left:
        current_min = current_min.left

    current_max = root
    while current_max.right:
        current_max = current_max.right

    return current_min.val * current_max.val

arr = list(map(int, input("Введите элементы BST через пробел: ").split()))
root = build_tree(arr)
result = find_min_max_product(root)
print(f"Произведение мин и макс элементов: {result}")
```

В бинарном дереве поиска (BST) действует правило: все значения в левом поддереве меньше корня, а в правом — больше.
Поэтому, чтобы найти минимум, мы просто идем от корня влево до упора. Чтобы найти максимум — идем вправо до упора.
Длина этого пути равна высоте дерева h. Поэтому поиск занимает O(h), что в сбалансированном дереве равно O(log n).

# Задача 5. Сравнение двух бинарных деревьев

## Условие
Даны два бинарных дерева. Необходимо определить, являются ли они идентичными (по структуре и значениям узлов).

**Что нужно сделать:**
Реализуйте рекурсивную проверку соответствия узлов обоих деревьев. Объясните, какие условия должны выполняться, чтобы деревья считались одинаковыми.

```python
def is_same_tree(p, q):
    if not p and not q:
        return True

    if not p or not q:
        return False

    if p.val != q.val:
        return False

    return is_same_tree(p.left, q.left) and is_same_tree(p.right, q.right)

print("Дерево 1:")
arr1 = list(map(int, input("Введите элементы первого дерева: ").split()))
root1 = build_tree(arr1)

print("Дерево 2:")
arr2 = list(map(int, input("Введите элементы второго дерева: ").split()))
root2 = build_tree(arr2)

result = is_same_tree(root1, root2)
print(f"Деревья идентичны: {result}")
```

**Объяснение:**
Деревья считаются одинаковыми, если:
1. Оба текущих узла пусты (None).
2. Значения в узлах равны.
3. Левые поддеревья одинаковы.
4. Правые поддеревья одинаковы.
Мы проверяем эти условия рекурсивно для каждого узла. Сложность алгоритма \(O(n)\), так как в худшем случае нам придется посетить все узлы обоих деревьев.
